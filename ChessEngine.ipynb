```python
class Pawn_Movements:
    def __init__(self, board):
        position = board.eq("♟1").stack()
        position = position[position].index[0]

        self.row, self.column = position

    def show_move(self):
        if self.column == "B":
            print("Pawn can move 1 or 2 spaces")

            move = input("Choose 1 or 2: ")

            if move == "1":
                new_row = self.row + 1
                self.move_pawn(new_row)

            elif move == "2":
                new_row = self.row + 2
                self.move_pawn(new_row)

        else:
            print("Pawn can move 1 space")

            new_row = self.row + 1
            self.move_pawn(new_row)

    def move_pawn(self, new_row):
        # Remove pawn from old position
        self.board.loc[self.row, self.column] = " "

        # Put pawn in new position

        self.board.loc[new_row, self.column] = "♟1"

        # Update pawn's position
        self.row = new_row
```
